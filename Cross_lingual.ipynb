```python
# --------------------------
# Veri Tanımlaması
# --------------------------
gruplar = ["Takım A (Hibrit)", "Takım B (Geleneksel Ofis)"]
ortalama_puanlar = [73.5, 65.0]
# %95 Güven Aralığı değerleri (Alt Sınır, Üst Sınır)
guven_araliklari = [
    [73.5 - 71, 76 - 73.5], # Takım A: Alt hata miktarı, Üst hata miktarı
    [65.0 - 63, 67 - 65.0]  # Takım B: Alt hata miktarı, Üst hata miktarı
]

# --------------------------
# Grafik Çizimi
# --------------------------
plt.figure(figsize=(7, 6))

# Bar grafiği + hata çubukları
barlar = plt.bar(
    gruplar,
    ortalama_puanlar,
    color = ["#2563eb", "#dc2626"],
    width = 0.5,
    yerr = np.array(guven_araliklari).T, # Hata çubukları için veriyi matplotlib formatına çevir
    capsize = 12 # Hata çubuğu kenar çizgisi boyutu
)

# Barların üstüne ortalama ve güven aralığı bilgisini yazdır
for bar, ortalama, alt_ga, ust_ga in zip(barlar, ortalama_puanlar, [71, 63], [76, 67]):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.4,
        f"Ortalama: {ortalama}\n%95 GA: {alt_ga} - {ust_ga}",
        ha = "center",
        fontweight = "bold"
    )

# Ayarlar
plt.title("Ortalama Verimlilik ve %95 Güven Aralıkları", fontsize=15, pad=20)
plt.ylabel("Ortalama Verimlilik Puanı", fontsize=12)
plt.ylim(60, 80)

plt.show()
```

```python
# =========================================================================================
# FINAL STABLE END-TO-END PIPELINE (All Fixes Applied)
# =========================================================================================

# 1. KURULUM
!pip -q install datasets pandas openpyxl scikit-learn sentence-transformers transformers accelerate sacremoses sentencepiece evaluate hangul-romanize

import os, re, random, gc
import numpy as np
import pandas as pd
import torch
from google.colab import drive
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score, precision_score, recall_score, precision_recall_fscore_support
from datasets import Dataset, load_dataset
from transformers import (AutoTokenizer, AutoModelForSequenceClassification, AutoModelForSeq2SeqLM,
                          TrainingArguments, Trainer, DataCollatorWithPadding)
from sentence_transformers import SentenceTransformer
from sklearn.linear_model import LogisticRegression
from hangul_romanize import Transliter
from hangul_romanize.rule import academic

# 2. AYARLAR
drive.mount('/content/drive')
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
BASE_DIR = "/content/drive/MyDrive/SIGTURK"
OUT_DIR  = f"{BASE_DIR}/FINAL_RUN_V3"
CKPT_DIR = f"{BASE_DIR}/checkpoints/xlmr_tr_final_v3"
MODEL_NAME = "xlm-roberta-base"

os.makedirs(OUT_DIR, exist_ok=True)
os.makedirs(CKPT_DIR, exist_ok=True)

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

def clear_gpu():
    for var in ['trainer', 'model', 'tokenizer', 'mdl_mt', 'tok_mt', 'mdl_tr', 'tok_tr', 'temp_mdl', 'temp_trainer', 'mdl_a', 'tok_a']:
        if var in globals():
            del globals()[var]
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

def eval_metrics(y_true, y_pred):
    return {
        "acc": accuracy_score(y_true, y_pred),
        "macro_f1": f1_score(y_true, y_pred, average="macro"),
        "f1_off": f1_score(y_true, y_pred, pos_label=1, zero_division=0),
        "prec_m": precision_score(y_true, y_pred, average="macro", zero_division=0),
        "rec_m": recall_score(y_true, y_pred, average="macro", zero_division=0),
    }

@torch.no_grad()
def predict(model, tokenizer, df, batch_size=32, max_len=128):
    model.eval()
    texts = df["text"].astype(str).tolist()
    preds = []
    for i in range(0, len(texts), batch_size):
        batch = texts[i:i+batch_size]
        enc = tokenizer(batch, return_tensors="pt", padding=True, truncation=True, max_length=max_len).to(DEVICE)
        logits = model(**enc).logits
        preds.extend(torch.argmax(logits, dim=-1).detach().cpu().numpy().tolist())
    return np.array(preds)

set_seed(42)

# 3. VERİ YÜKLEME
print("📥 Veriler yükleniyor...")
df_tr_raw = pd.read_excel(f"{BASE_DIR}/ATC.xlsx")
df_tr = df_tr_raw[["message","label"]].rename(columns={"message":"text"}).dropna()
df_tr["label"] = df_tr["label"].astype(int)
df_tr_train, df_tr_tmp = train_test_split(df_tr, test_size=0.2, random_state=42, stratify=df_tr["label"])
df_tr_dev, df_tr_test  = train_test_split(df_tr_tmp, test_size=0.5, random_state=42, stratify=df_tr_tmp["label"])

ko_test_raw = pd.read_csv("https://raw.githubusercontent.com/kocohub/korean-hate-speech/master/labeled/dev.tsv", sep="\t")
df_ko_test = ko_test_raw.assign(label=ko_test_raw["hate"].apply(lambda x: 0 if x=="none" else 1))[["comments","label"]].rename(columns={"comments":"text"})

df_en_test = pd.DataFrame(load_dataset("tweet_eval", "offensive", split="test"))[["text","label"]]

hu_raw = pd.DataFrame(load_dataset("RabidUmarell/hungarian-toxic-comments", split="train"))
hu_raw["label"] = (hu_raw["toxic"].astype(float) >= 2.5).astype(int)
_, df_hu_test = train_test_split(hu_raw[["text","label"]], test_size=0.5, random_state=42, stratify=hu_raw["label"])

fi_raw = pd.read_excel(f"{BASE_DIR}/RTP_LX_FI.xlsx")
df_fi_test_full = fi_raw.rename(columns={"Completion":"text"}).copy()
df_fi_test_full["label"] = (df_fi_test_full["CompletionAnnotations.Toxicity"].astype(float) >= 3.5).astype(int)
_, df_fi_test = train_test_split(df_fi_test_full[["text","label"]], test_size=0.4, random_state=42, stratify=df_fi_test_full["label"])

def get_sub(df, n=200):
    return df.groupby("label", group_keys=False).apply(lambda x: x.sample(min(len(x), n//2), random_state=42))

sub_ko, sub_fi, sub_hu, sub_en = get_sub(df_ko_test), get_sub(df_fi_test), get_sub(df_hu_test), get_sub(df_en_test)
print("✓ Veri yükleme tamam.")

# 4. ANA TR MODEL EĞİTİMİ
print("🏋️ Ana TR modeli eğitiliyor...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2).to(DEVICE)
def tok_fn(x): return tokenizer(x["text"], truncation=True, max_length=128)
ds_train = Dataset.from_pandas(df_tr_train).map(tok_fn, batched=True)
ds_dev   = Dataset.from_pandas(df_tr_dev).map(tok_fn, batched=True)

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    p, r, f1, _ = precision_recall_fscore_support(labels, preds, average="macro", zero_division=0)
    return {"accuracy": accuracy_score(labels, preds), "macro_f1": f1}

args = TrainingArguments(output_dir=CKPT_DIR, learning_rate=2e-5, per_device_train_batch_size=16,
                         num_train_epochs=3, eval_strategy="epoch", save_strategy="epoch",
                         load_best_model_at_end=True, metric_for_best_model="macro_f1", report_to="none")

trainer = Trainer(model=model, args=args, train_dataset=ds_train, eval_dataset=ds_dev,
                  data_collator=DataCollatorWithPadding(tokenizer), compute_metrics=compute_metrics)
trainer.train()
trainer.save_model(CKPT_DIR)
tokenizer.save_pretrained(CKPT_DIR)
mdl_tr, tok_tr = model, tokenizer
print("✓ Ana model kaydedildi.")

# 5. ZERO-SHOT & LaBSE & FRAG & ROMANIZATION
print("📊 Zero-shot ve Baselines başlıyor...")
zs_results = []
for n, d in [("KO", sub_ko), ("FI", sub_fi), ("HU", sub_hu), ("EN", sub_en)]:
    p = predict(mdl_tr, tok_tr, d)
    zs_results.append({"exp":"ZeroShot", "setting":n, **eval_metrics(d["label"].values, p)})
df_zs = pd.DataFrame(zs_results)

labse = SentenceTransformer('sentence-transformers/LaBSE')
def eval_labse(train_df, test_df, name):
    Xtr, Xte = labse.encode(train_df["text"].tolist()), labse.encode(test_df["text"].tolist())
    clf = LogisticRegression(max_iter=1000, class_weight="balanced").fit(Xtr, train_df["label"])
    return {"exp":"LaBSE", "setting":name, **eval_metrics(test_df["label"].values, clf.predict(Xte))}
df_b1 = pd.DataFrame([eval_labse(df_tr_train.sample(2000), d, n) for n, d in [("KO", sub_ko), ("FI", sub_fi), ("HU", sub_hu), ("EN", sub_en)]])

def get_frag(df, name):
    t_cnt = [len(tok_tr.tokenize(str(t))) for t in df["text"]]
    w_cnt = [len(str(t).split()) for t in df["text"]]
    return {"lang": name, "ratio": np.mean(t_cnt)/np.mean(w_cnt)}
df_frag = pd.DataFrame([get_frag(d, n) for n, d in [("TR", df_tr_test), ("KO", sub_ko), ("FI", sub_fi), ("HU", sub_hu), ("EN", sub_en)]])

trans = Transliter(academic)
sub_ko_rom = sub_ko.copy(); sub_ko_rom["text"] = sub_ko_rom["text"].astype(str).apply(lambda x: trans.translit(x))
df_rom = pd.DataFrame([{"exp":"Romanization", "setting":"KO_ROM", **eval_metrics(sub_ko_rom["label"].values, predict(mdl_tr, tok_tr, sub_ko_rom))}])

# 6. BASELINE-2: TRANSLATE-TEST (NLLB)
print("🌐 NLLB Çeviri Baseline başlıyor...")
clear_gpu()
mt_tok = AutoTokenizer.from_pretrained("facebook/nllb-200-distilled-600M")
mt_mdl = AutoModelForSeq2SeqLM.from_pretrained("facebook/nllb-200-distilled-600M").to(DEVICE)

def translate(texts, lang):
    mt_tok.src_lang = lang
    forced_bos = mt_tok.convert_tokens_to_ids("tur_Latn")
    inputs = mt_tok(texts, return_tensors="pt", padding=True, truncation=True).to(DEVICE)
    gen = mt_mdl.generate(**inputs, forced_bos_token_id=forced_bos, max_new_tokens=100)
    return mt_tok.batch_decode(gen, skip_special_tokens=True)

b2_res = []
mdl_tr_reloaded = AutoModelForSequenceClassification.from_pretrained(CKPT_DIR).to(DEVICE)
for d, code, n in [(sub_ko,"kor_Hang","KO"), (sub_fi,"fin_Latn","FI"), (sub_hu,"hun_Latn","HU"), (sub_en,"eng_Latn","EN")]:
    tr_texts = translate(d["text"].tolist(), code)
    df_eval = pd.DataFrame({"text": tr_texts, "label": d["label"].values})
    p = predict(mdl_tr_reloaded, tok_tr, df_eval)
    b2_res.append({"exp":"Translate-Test", "setting":n, **eval_metrics(d["label"].values, p)})
df_b2 = pd.DataFrame(b2_res)

# 7. ABLATIONS
print("🔬 Ablation çalışmaları başlıyor...")
def train_abl(train_df, m_len, name):
    clear_gpu()
    t_mdl = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2).to(DEVICE)
    ds = Dataset.from_pandas(train_df).map(lambda x: tok_tr(x["text"], truncation=True, max_length=m_len), batched=True)
    t_args = TrainingArguments(output_dir="./tmp", num_train_epochs=1, per_device_train_batch_size=16, save_strategy="no", report_to="none")
    t = Trainer(model=t_mdl, args=t_args, train_dataset=ds, data_collator=DataCollatorWithPadding(tok_tr))
    t.train()
    res = []
    for sub, n in [(sub_ko,"KO"), (sub_fi,"FI"), (sub_hu,"HU"), (sub_en,"EN")]:
        p = predict(t_mdl, tok_tr, sub, max_len=m_len)
        res.append({"exp":"Ablation", "setting":f"{name}_L{m_len}_{n}", **eval_metrics(sub["label"].values, p)})
    return res

a1_res = train_abl(df_tr_train.sample(5000, random_state=42), 128, "Size5k")
df_abl = pd.DataFrame(a1_res)

# 8. FINAL RAPORLAMA
print("📝 Sonuçlar birleştiriliyor...")
df_final = pd.concat([df_zs, df_b1, df_b2, df_rom, df_abl], ignore_index=True)
df_final.to_excel(f"{OUT_DIR}/SIGTURK_FULL_REPORT.xlsx", index=False)
pivot = df_final.pivot_table(index="setting", columns="exp", values="macro_f1", aggfunc="first")
pivot.to_excel(f"{OUT_DIR}/PAPER_TABLE_MACRO_F1.xlsx")
df_frag.to_excel(f"{OUT_DIR}/TOKEN_FRAG_ANALYSIS.xlsx", index=False)

print(f"✅ TÜM SÜREÇ TAMAMLANDI. Raporlar: {OUT_DIR}")
pivot
```

```python
# ============================================================================
# KURTARMA VE FİNALİZASYON KODU (OOM KESİN ÇÖZÜM)
# ============================================================================
import gc, torch, pandas as pd
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, AutoModelForSequenceClassification

# 1. BELLEK DOSTU NLLB YÜKLEME
print("📥 NLLB-200 yükleniyor (FP16 Mode)...")
mt_name = "facebook/nllb-200-distilled-600M"
mt_tok = AutoTokenizer.from_pretrained(mt_name)
# torch_dtype=torch.float16 belleği %50 azaltır
mt_mdl = AutoModelForSeq2SeqLM.from_pretrained(mt_name, torch_dtype=torch.float16).to(DEVICE)
mt_mdl.eval()

@torch.no_grad()
def fast_translate_v3(texts, src_lang):
    mt_tok.src_lang = src_lang
    forced_bos = mt_tok.convert_tokens_to_ids("tur_Latn")
    outs = []
    for i in range(0, len(texts), 8): # Küçük batch size
        batch = [str(t)[:512] for t in texts[i:i+8]]
        enc = mt_tok(batch, return_tensors="pt", padding=True, truncation=True, max_length=256).to(DEVICE)
        gen = mt_mdl.generate(**enc, forced_bos_token_id=forced_bos, max_new_tokens=100)
        outs.extend(mt_tok.batch_decode(gen, skip_special_tokens=True))
    return outs

# Çevirileri Yap
all_translated_dfs = {}
jobs = [(sub_ko, "kor_Hang", "KO"), (sub_fi, "fin_Latn", "FI"), (sub_hu, "hun_Latn", "HU"), (sub_en, "eng_Latn", "EN")]

for df_sub, code, name in jobs:
    print(f"🚀 {name} çevriliyor...")
    # DataFrame/Series çakışma koruması
    raw_list = df_sub["text"].iloc[:, 0].tolist() if isinstance(df_sub["text"], pd.DataFrame) else df_sub["text"].tolist()
    tr_list = fast_translate_v3(raw_list, code)
    all_translated_dfs[name] = pd.DataFrame({"text": tr_list, "label": df_sub["label"].values})

# NLLB'yi SİL
del mt_mdl, mt_tok
gc.collect()
torch.cuda.empty_cache()

# 2. ANA MODELİ YÜKLE VE TEST ET
print(f"📥 Ana model yükleniyor: {CKPT_DIR}")
tok_tr = AutoTokenizer.from_pretrained(CKPT_DIR)
mdl_tr_reloaded = AutoModelForSequenceClassification.from_pretrained(CKPT_DIR).to(DEVICE)

b2_res = []
for name, df_eval in all_translated_dfs.items():
    print(f"📊 {name} test ediliyor...")
    # predict fonksiyonun bir önceki başarılı hücrede tanımlı olmalı
    p = predict(mdl_tr_reloaded, tok_tr, df_eval)
    b2_res.append({"exp":"Translate-Test", "setting":name, **eval_metrics(df_eval["label"].values, p)})

df_b2 = pd.DataFrame(b2_res)

# 3. YEDEKLEDİĞİN VERİLERİ GERİ ÇEK VE BİRLEŞTİR
print("📝 Final raporlar birleştiriliyor...")
try:
    df_zs_old = pd.read_excel(f"{OUT_DIR}/backup_zs.xlsx")
    df_b1_old = pd.read_excel(f"{OUT_DIR}/backup_b1.xlsx")
    df_rom_old = pd.read_excel(f"{OUT_DIR}/backup_rom.xlsx")

    df_final = pd.concat([df_zs_old, df_b1_old, df_b2, df_rom_old], ignore_index=True)
    df_final.to_excel(f"{OUT_DIR}/SIGTURK_FINAL_REPORT_COMPLETED.xlsx", index=False)

    pivot = df_final.pivot_table(index="setting", columns="exp", values="macro_f1", aggfunc="first")
    pivot.to_excel(f"{OUT_DIR}/PAPER_TABLE_MACRO_F1_FINAL.xlsx")
    print(f"🏁 TEBRİKLER! Tüm süreç bitti. Sonuçlar: {OUT_DIR}")
    display(pivot)
except Exception as e:
    print(f"⚠️ Rapor birleştirme hatası (Yedekler bulunamadı?): {e}")
    print("Ancak Baseline-2 (Çeviri) başarıyla tamamlandı, manuel birleştirebilirsiniz.")
    display(df_b2)
```

```python
# =========================================================================================
# FINAL SOTA PIPELINE - VERSION 4 (Stable & Comprehensive)
# =========================================================================================

# 1. KURULUM (Laser kaldırıldı, LaBSE ve diğerleri korundu)
!pip -q install datasets pandas openpyxl scikit-learn sentence-transformers transformers accelerate \
                sacremoses sentencepiece evaluate hangul-romanize

import os, re, random, gc
import numpy as np
import pandas as pd
import torch
from google.colab import drive
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score, precision_recall_fscore_support
from datasets import Dataset, load_dataset
from transformers import (AutoTokenizer, AutoModelForSequenceClassification, AutoModelForSeq2SeqLM,
                          TrainingArguments, Trainer, DataCollatorWithPadding)
from sentence_transformers import SentenceTransformer
from sklearn.linear_model import LogisticRegression
from hangul_romanize import Transliter
from hangul_romanize.rule import academic

# 2. AYARLAR
drive.mount('/content/drive')
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
BASE_DIR = "/content/drive/MyDrive/SIGTURK"
OUT_DIR  = f"{BASE_DIR}/SOTA_FINAL_RESULTS_V4"
CKPT_DIR = f"{BASE_DIR}/checkpoints/xlmr_final_stable"
MODEL_NAME = "xlm-roberta-base"

os.makedirs(OUT_DIR, exist_ok=True)
os.makedirs(CKPT_DIR, exist_ok=True)

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

def clear_gpu():
    global trainer, model, mdl_tr_reloaded, mt_mdl, mt_tok, labse
    for var in ['trainer', 'model', 'mdl_tr_reloaded', 'mt_mdl', 'mt_tok', 'temp_mdl', 't_trainer']:
        if var in globals(): del globals()[var]
    gc.collect()
    if torch.cuda.is_available(): torch.cuda.empty_cache(); torch.cuda.ipc_collect()

def basic_clean(text):
    t = str(text)
    t = re.sub(r"http\S+|www\.\S+", " ", t); t = re.sub(r"@\w+", " ", t)
    t = re.sub(r"#\w+", " ", t); t = re.sub(r"\s+", " ", t).strip()
    return t

def eval_metrics(y_true, y_pred):
    p, r, f1, _ = precision_recall_fscore_support(y_true, y_pred, average="macro", zero_division=0)
    return {"acc": accuracy_score(y_true, y_pred), "macro_f1": f1, "prec_m": p, "rec_m": r}

@torch.no_grad()
def predict_fast(model, tokenizer, df, max_len=128):
    model.eval()
    texts = df["text"].astype(str).tolist()
    enc = tokenizer(texts, return_tensors="pt", padding=True, truncation=True, max_length=max_len).to(DEVICE)
    logits = model(**enc).logits
    return torch.argmax(logits, dim=-1).cpu().numpy()

set_seed(42)

# 3. VERİ YÜKLEME
print("📥 Veriler hazırlanıyor...")
df_tr = pd.read_excel(f"{BASE_DIR}/ATC.xlsx")[["message","label"]].rename(columns={"message":"text"}).dropna()
df_tr["label"] = df_tr["label"].astype(int)
df_tr_train, df_tr_dev = train_test_split(df_tr, test_size=0.1, random_state=42, stratify=df_tr["label"])

ko_raw = pd.read_csv("https://raw.githubusercontent.com/kocohub/korean-hate-speech/master/labeled/dev.tsv", sep="\t")
df_ko_test = ko_raw.assign(label=ko_raw["hate"].apply(lambda x: 0 if x=="none" else 1))[["comments","label"]].rename(columns={"comments":"text"})
df_en_test = pd.DataFrame(load_dataset("tweet_eval", "offensive", split="test"))[["text","label"]]
hu_raw = pd.DataFrame(load_dataset("RabidUmarell/hungarian-toxic-comments", split="train"))
hu_raw["label"] = (hu_raw["toxic"].astype(float) >= 2.5).astype(int)
_, df_hu_test = train_test_split(hu_raw[["text","label"]], test_size=0.5, random_state=42, stratify=hu_raw["label"])
fi_raw = pd.read_excel(f"{BASE_DIR}/RTP_LX_FI.xlsx").rename(columns={"Completion":"text"})
fi_raw["label"] = (fi_raw["CompletionAnnotations.Toxicity"].astype(float) >= 3.5).astype(int)
_, df_fi_test = train_test_split(fi_raw[["text","label"]], test_size=0.4, random_state=42, stratify=fi_raw["label"])

# Hizli alt kumeler (N=200)
def get_sub(df, n=200): return df.groupby("label", group_keys=False).apply(lambda x: x.sample(min(len(x), n//2), random_state=42))
sub_ko, sub_fi, sub_hu, sub_en = get_sub(df_ko_test), get_sub(df_fi_test), get_sub(df_hu_test), get_sub(df_en_test)

# =========================================================================================
# 4. INSTANCE TRANSFER (Çeviri Deneyleri)
# =========================================================================================
print("🌐 NLLB-200 Deneyleri (Translate-Test & Back-Translation)...")
clear_gpu()
mt_name = "facebook/nllb-200-distilled-600M"
mt_tok = AutoTokenizer.from_pretrained(mt_name)
mt_mdl = AutoModelForSeq2SeqLM.from_pretrained(mt_name, torch_dtype=torch.float16).to(DEVICE)

def translate_v4(texts, src, tgt="tur_Latn"):
    mt_tok.src_lang = src
    forced_bos = mt_tok.convert_tokens_to_ids(tgt)
    outs = []
    for i in range(0, len(texts), 8):
        batch = [str(t)[:512] for t in texts[i:i+8]]
        enc = mt_tok(batch, return_tensors="pt", padding=True, truncation=True).to(DEVICE)
        gen = mt_mdl.generate(**enc, forced_bos_token_id=forced_bos, max_new_tokens=100)
        outs.extend(mt_tok.batch_decode(gen, skip_special_tokens=True))
    return outs

# 4.1 Translate-Test
trans_test_results = {}
for df_s, code, name in [(sub_ko,"kor_Hang","KO"), (sub_fi,"fin_Latn","FI"), (sub_hu,"hun_Latn","HU"), (sub_en,"eng_Latn","EN")]:
    print(f"🚀 Translate-Test: {name}")
    trans_test_results[name] = pd.DataFrame({"text": translate_v4(df_s["text"].tolist(), code), "label": df_s["label"].values})

# 4.2 Back-Translation (Augmentation)
print("🚀 Back-Translation (TR->EN->TR)...")
tr_sample = df_tr_train.sample(400, random_state=42)
en_mid = translate_v4(tr_sample["text"].tolist(), "tur_Latn", "eng_Latn")
df_back_tr = pd.DataFrame({"text": translate_v4(en_mid, "eng_Latn", "tur_Latn"), "label": tr_sample["label"].values})

del mt_mdl, mt_tok; clear_gpu()

# =========================================================================================
# 5. MODEL EĞİTİMİ (XLM-R)
# =========================================================================================
print("🏋️ Ana Model ve Ablationlar...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def run_train(train_df, m_len=128, epochs=3):
    clear_gpu()
    m = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2).to(DEVICE)
    ds = Dataset.from_pandas(train_df).map(lambda x: tokenizer(x["text"], truncation=True, max_length=m_len), batched=True)
    args = TrainingArguments(output_dir="./tmp", num_train_epochs=epochs, per_device_train_batch_size=16, report_to="none", save_strategy="no", fp16=True)
    t = Trainer(model=m, args=args, train_dataset=ds, data_collator=DataCollatorWithPadding(tokenizer))
    t.train()
    return m

# --- 5.1 Ana Model ---
mdl_tr = run_train(df_tr_train.sample(8000, random_state=42)) # Hızlı eğitim için 8k

# Zero-Shot Değerlendirme
zs_res = []
for n, d in [("KO", sub_ko), ("FI", sub_fi), ("HU", sub_hu), ("EN", sub_en)]:
    zs_res.append({"exp":"ZeroShot", "setting":n, **eval_metrics(d["label"].values, predict_fast(mdl_tr, tokenizer, d))})

# Translate-Test Değerlendirme
tt_res = []
for n, d in trans_test_results.items():
    tt_res.append({"exp":"Translate-Test", "setting":n, **eval_metrics(d["label"].values, predict_fast(mdl_tr, tokenizer, d))})

# --- 5.2 Frozen Encoder Baseline ---
print("❄️ Frozen Encoder Deneyi...")
mdl_frozen = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2).to(DEVICE)
for param in mdl_frozen.roberta.parameters(): param.requires_grad = False
# ... (Kısa eğitim)
mdl_frozen = run_train(df_tr_train.sample(3000), epochs=1)
frozen_res = [{"exp":"Frozen", "setting":n, **eval_metrics(d["label"].values, predict_fast(mdl_frozen, tokenizer, d))} for n, d in [("KO", sub_ko), ("EN", sub_en)]]

# =========================================================================================
# 6. FEATURE TRANSFER & LINGUISTIC
# =========================================================================================
print("📊 LaBSE & Fragmentation...")
labse = SentenceTransformer('sentence-transformers/LaBSE')
def get_labse(tr_df, te_df, name):
    clf = LogisticRegression(max_iter=1000).fit(labse.encode(tr_df["text"].tolist()), tr_df["label"])
    return {"exp":"LaBSE", "setting":name, **eval_metrics(te_df["label"].values, clf.predict(labse.encode(te_df["text"].tolist())))}
b1_res = [get_labse(df_tr_train.sample(2000), d, n) for n, d in [("KO", sub_ko), ("EN", sub_en)]]

# Fragmentation Ratio
frag_stats = [{"lang": n, "ratio": np.mean([len(tokenizer.tokenize(str(t))) for t in d["text"]]) / np.mean([len(str(t).split()) for t in d["text"]])}
              for n, d in [("TR", df_tr_test.sample(500)), ("KO", sub_ko), ("FI", sub_fi), ("HU", sub_hu), ("EN", sub_en)]]

# =========================================================================================
# 7. RAPORLAMA
# =========================================================================================
df_final = pd.concat([pd.DataFrame(zs_res), pd.DataFrame(tt_res), pd.DataFrame(frozen_res), pd.DataFrame(b1_res)], ignore_index=True)
df_final.to_excel(f"{OUT_DIR}/SOTA_FINAL_V4.xlsx", index=False)
pd.DataFrame(frag_stats).to_excel(f"{OUT_DIR}/FRAGMENTATION_V4.xlsx", index=False)

print(f"✅ BİTTİ. Raporlar: {OUT_DIR}")
df_final.pivot_table(index="setting", columns="exp", values="macro_f1")
```
